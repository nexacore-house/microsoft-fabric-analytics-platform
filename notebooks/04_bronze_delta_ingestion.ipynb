# Bronze Delta Ingestion

Ingests the full partitioned London Smart Meter dataset from the OneLake landing zone into the Bronze layer of the Fabric Lakehouse.

## Purpose

- Read all 168 landed smart-meter CSV files
- Apply an explicit raw-source schema
- Preserve source values with minimal transformation
- Normalise the source consumption column name for Delta compatibility
- Add file and ingestion lineage metadata
- Persist the full dataset as a Bronze Delta table
- Reconcile Bronze record and source-file counts against the landing zone

> Bronze preserves the source data at its lowest practical level of transformation. Data-quality corrections and deduplication are intentionally deferred to the Silver layer.

```python
# ----------------------------------------
# Project 05 - Microsoft Fabric Analytics Platform
# Notebook 05 - Bronze Delta Ingestion
# ----------------------------------------

from pyspark.sql import functions as F
from pyspark.sql.types import (
    StructType,
    StructField,
    StringType
)

source_path = (
    "Files/landing/meter_readings/"
    "Small LCL Data/*.csv"
)

print("Bronze Delta ingestion notebook initialised.")
```

## 1. Define Landing Source and Raw Schema

```python
bronze_schema = StructType([
    StructField("LCLid", StringType(), True),
    StructField("stdorToU", StringType(), True),
    StructField("DateTime", StringType(), True),
    StructField("KWH/hh (per half hour) ", StringType(), True)
])

print("Explicit Bronze source schema defined.")
```

## 2. Read Full Landing Dataset

```python
landing_df = (
    spark.read
    .option("header", True)
    .schema(bronze_schema)
    .csv(source_path)
)

print("Landing files loaded into Spark.")
```

## 3. Apply Bronze Column Normalisation

```python
bronze_df = (
    landing_df
    .withColumn(
        "SourceFilePath",
        F.input_file_name()
    )
    .withColumn(
        "SourceFileName",
        F.regexp_extract(
            F.input_file_name(),
            r"([^/]+$)",
            1
        )
    )
    .withColumn(
        "IngestionTimestamp",
        F.current_timestamp()
    )
    .withColumn(
        "SourceSystem",
        F.lit("London Datastore")
    )
)
```

```python
display(
    bronze_df
    .select(
        "LCLid",
        "stdorToU",
        "DateTime",
        "KWH/hh (per half hour) ",
        "SourceFileName",
        "SourceSystem"
    )
    .limit(10)
)
```

```python
spark.sql(
    "CREATE SCHEMA IF NOT EXISTS bronze"
)

print("Bronze schema ready.")
```

### Delta-Compatible Column Naming

The source field `KWH/hh (per half hour) ` contains characters that are unsuitable for the target Delta schema.

The field is renamed to `ConsumptionKWhRaw` while its source value remains unchanged. This is a physical schema normalisation rather than a data-quality transformation.

```python
bronze_df = (
    landing_df
    .withColumnRenamed(
        "KWH/hh (per half hour) ",
        "ConsumptionKWhRaw"
    )
    .withColumn(
        "SourceFilePath",
        F.input_file_name()
    )
    .withColumn(
        "SourceFileName",
        F.regexp_extract(
            F.input_file_name(),
            r"([^/]+$)",
            1
        )
    )
    .withColumn(
        "IngestionTimestamp",
        F.current_timestamp()
    )
    .withColumn(
        "SourceSystem",
        F.lit("London Datastore")
    )
)
```

```python
(
    bronze_df
    .write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("bronze.meter_readings")
)

print("bronze.meter_readings written successfully.")
```
