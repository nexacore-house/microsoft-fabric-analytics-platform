```python
# ----------------------------------------
# Project 05 - Microsoft Fabric Analytics Platform
# Notebook 05 - Bronze Delta Ingestion
# ----------------------------------------

from pyspark.sql import functions as F
from pyspark.sql.types import (
    StructType,
    StructField,
    StringType
)

source_path = (
    "Files/landing/meter_readings/"
    "Small LCL Data/*.csv"
)

print("Bronze Delta ingestion notebook initialised.")
```

```python
bronze_schema = StructType([
    StructField("LCLid", StringType(), True),
    StructField("stdorToU", StringType(), True),
    StructField("DateTime", StringType(), True),
    StructField("KWH/hh (per half hour) ", StringType(), True)
])

print("Explicit Bronze source schema defined.")
```

```python
landing_df = (
    spark.read
    .option("header", True)
    .schema(bronze_schema)
    .csv(source_path)
)

print("Landing files loaded into Spark.")
```

```python
bronze_df = (
    landing_df
    .withColumn(
        "SourceFilePath",
        F.input_file_name()
    )
    .withColumn(
        "SourceFileName",
        F.regexp_extract(
            F.input_file_name(),
            r"([^/]+$)",
            1
        )
    )
    .withColumn(
        "IngestionTimestamp",
        F.current_timestamp()
    )
    .withColumn(
        "SourceSystem",
        F.lit("London Datastore")
    )
)
```

```python
display(
    bronze_df
    .select(
        "LCLid",
        "stdorToU",
        "DateTime",
        "KWH/hh (per half hour) ",
        "SourceFileName",
        "SourceSystem"
    )
    .limit(10)
)
```

```python
spark.sql(
    "CREATE SCHEMA IF NOT EXISTS bronze"
)

print("Bronze schema ready.")
```

```python
bronze_df = (
    landing_df
    .withColumnRenamed(
        "KWH/hh (per half hour) ",
        "ConsumptionKWhRaw"
    )
    .withColumn(
        "SourceFilePath",
        F.input_file_name()
    )
    .withColumn(
        "SourceFileName",
        F.regexp_extract(
            F.input_file_name(),
            r"([^/]+$)",
            1
        )
    )
    .withColumn(
        "IngestionTimestamp",
        F.current_timestamp()
    )
    .withColumn(
        "SourceSystem",
        F.lit("London Datastore")
    )
)
```

```python
(
    bronze_df
    .write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("bronze.meter_readings")
)

print("bronze.meter_readings written successfully.")
```
