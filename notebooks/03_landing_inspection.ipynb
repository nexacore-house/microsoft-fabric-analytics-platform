```python
from notebookutils import mssparkutils

landing_path = "Files/Small LCL Data"

files = mssparkutils.fs.ls(landing_path)

print("LANDING FILE INSPECTION")
print("-" * 50)
print(f"File count: {len(files)}")

for file in files[:10]:
    print(file.name, file.size)
```

```python
total_size_bytes = sum(file.size for file in files)

total_size_gb = total_size_bytes / (1024 ** 3)

print("LANDING SIZE")
print("-" * 50)
print(f"Files: {len(files)}")
print(f"Total size: {total_size_gb:.2f} GB")
```

```python
print("SAMPLE FILE NAMES")
print("-" * 50)

for file in files[:20]:
    print(file.name)
```

```python
sample_file = files[0].path

sample_df = (
    spark.read
    .option("header", True)
    .option("inferSchema", False)
    .csv(sample_file)
)

print("SAMPLE SCHEMA")
print("-" * 50)

sample_df.printSchema()

display(sample_df.limit(10))
```

```python
sample_paths = [file.path for file in files[:10]]

tariff_sample_df = (
    spark.read
    .option("header", True)
    .option("inferSchema", False)
    .csv(sample_paths)
)

print("TARIFF VALUES IN SAMPLE FILES")
print("-" * 50)

tariff_sample_df.groupBy("stdorToU").count().show(truncate=False)
```

```python
# Select files distributed across the full archive
test_indexes = [0, 20, 40, 60, 80, 100, 120, 140, 160, 167]

test_paths = [
    files[i].path
    for i in test_indexes
    if i < len(files)
]

tariff_test_df = (
    spark.read
    .option("header", True)
    .option("inferSchema", False)
    .csv(test_paths)
)

print("DISTRIBUTED TARIFF TEST")
print("-" * 50)

tariff_test_df.groupBy("stdorToU").count().show(truncate=False)
```
